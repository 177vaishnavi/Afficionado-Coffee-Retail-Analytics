```python
from google.colab import drive
import os
import shutil

drive.mount('/content/drive')

project_folder = '/content/drive/MyDrive/Afficionado_Coffee_Retail_Analytics'
os.makedirs(project_folder, exist_ok=True)

# Copy dataset and Python modules
files_to_copy = [
    'Afficionado Coffee Roasters.xlsx',
    'data_loader.py',
    'kpi_analysis.py',
    'time_analysis.py',
    'visualizer.py'
]

for file in files_to_copy:
    shutil.copy2('/content/' + file, project_folder)

# Copy charts and analysis outputs
for folder in ['charts', 'outputs']:
    shutil.copytree(
        '/content/' + folder,
        project_folder + '/' + folder,
        dirs_exist_ok=True
    )

print("Complete project copied to Google Drive!")
```
